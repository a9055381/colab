```python
import random

CHOICES = {
    0: "剪刀",
    1: "石頭",
    2: "布",
}

circle = len(CHOICES)
```

```python
def build_move_info(number: int) -> dict:
    """將 0-5 的整數轉換為包含組別、出拳代號與原始值的字典。"""
    if number not in range(0, 6):
        raise ValueError("請輸入整數 0-5（0、3:剪刀, 1、4:石頭, 2、5:布）")
    return {
        "group": number // circle,
        "move": number % circle,
        "raw": number,
    }
```

```python
def com_random() -> dict:
    """隨機產生電腦出拳資訊。"""
    com = random.randint(0, 5)
    return build_move_info(com)
```

```python
def assess_rps(player: dict, com: dict) -> tuple[str, str]:
    """判斷勝負，回傳 (結果代碼, 提示訊息)。"""
    if player["group"] == com["group"]:
        if (com["move"] + 1) % circle == player["move"]:
            return "win", "你贏了，真的很厲害！"
        if (player["move"] + 1) % circle == com["move"]:
            return "lost", "電腦贏了，再接再厲！"
    return "even", "兩方平手，沒輸沒贏哦！"
```

```python
def play_game() -> None:
    """執行剪刀石頭布遊戲，直到分出勝負。"""
    while True:
        raw = input("請輸入剪刀(0、3)、石頭(1、4)、布(2、5)：").strip()
        try:
            player_info = build_move_info(int(raw))
            com_info = com_random()

            print(f"你在第{player_info['group']}組，出了{CHOICES[player_info['move']]} ({player_info['raw']})")
            print(f"電腦在第{com_info['group']}組，出了{CHOICES[com_info['move']]} ({com_info['raw']})")

            result, message = assess_rps(player_info, com_info)
            print(message)

            if result != "even":
                return
        except ValueError as exc:
            msg = str(exc) if str(exc) else "請輸入數字 0-5"
            print(f"輸入無效：{msg}，請再試一次。")
```

```python
if __name__ == "__main__":
    play_game()
```
