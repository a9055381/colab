```python
# replace 用法
str_data = "hello"
print(str_data)
str_data = str_data.replace("l","x")
print(str_data)
str_data = str_data.replace("x","p",1)
print(str_data)

# split 用法
str_data = "a,b,c"
date_data = "2026-11-23"
print(str_data)
print(date_data)
str_data = str_data.split(",")
date_data = date_data.split("-")
print(date_data)
str_data2 = "a905538@gmeil.com"

# join 用法
str_list = ["a", "b", "c"]
str_touple=("a", "b", "c")
print(str_list)
str_data = "-".join(str_list)
print(str_data)
print("/".join(str_touple))

# strip 用法
str_data = " hello "
print(str_data)
str_data = str_data.strip()
print(str_data)

# 查詢 string 所有專屬功能
for n in dir(str):
    print(n)
```

```python
list_data = [70, 20, 33]
list_data2 = list_data.append(["33", 40])
print(list_data)
print(list_data2)
print(list_data2 is None)
print(list_data2 == None)
print(type(list_data))
print(type(list_data2))
```

```python
# 所謂 print 沒有回傳值，是因為不能執行賦予(和 append 一樣)
result = print(30)
# print(30) 無法傳遞給 result，但是卻可以執行函式
print(result)
```

```python
# 錯誤寫法
list_data = [70, 20, 33]
list_data = list_data.append(["33", 40])
# 右側 list_data 執行 append 函式後，但是不會有回傳值，所以賦予 list_data 為 None
print(list_data)
```

```python
# insert 用法 .insert(插入位置, 數值)
list_data = [70, 20, 33]
list_data.insert(1, 50)
print(list_data)
```

```python
# pop 用法 .pop(位置/省略為取最尾值)
list_data1 = [70, 20, 33]
list_data2 = list_data1.pop()
print(list_data1)
print(list_data2)

list_data3 = list_data1.pop(0)
print(list_data1)
print(list_data3)
```
