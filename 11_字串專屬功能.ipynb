```python
# choice: int：參數型別標註（Type Hint），表示 choice 預期傳入整數（int）。
# -> int：回傳值型別標註，表示這個函式正常執行完畢後，預期回傳一個整數（int）。
# 利用 Python 的變數作用域（Scope）規則（常見的 LEGB 順序：Local → Enclosing → Global → Built-in）。
# 函式內部可以讀取外部參數（CHOICES 定義在檔案最外層，屬於 Global（全域） 變數）

def check_my_rps(choice: int) -> int:
    # 驗證出拳是否為 0, 1, 2；不合法則拋出 ValueError。
    # not in 判斷 choice 是否不在 CHOICES 的鍵（key：0, 1, 2）之中
    if choice not in CHOICES:

        #raise 的意思是主動拋出例外，中斷目前函式的正常往下執行，把錯誤往上丟給呼叫者處理。
        raise ValueError("請輸入整數 0-2（0:剪刀, 1:石頭, 2:布）")


    return choice
```

```python
def get_player_choice() -> int:
    # while True:無止盡迴圈，以下是停止迴圈的節點
    # 執行到 return check_my_rps(value) 時，會直接結束 get_player_choice() 函式並回傳結果，迴圈也跟著結束。
    # 如果改寫成用 break 跳出迴圈，也會停止。
    # 發生未捕捉的例外或程式被強制中斷：例如在終端機按 Ctrl + C 產生 KeyboardInterrupt，或遇到沒有被 except 接住的錯誤。

    while True:
        # .strip() 是字串（str）的方法，用來移除字串前後多餘的空白字元（包含空白鍵 ' '、換行 \n、Tab \t 等），但不會動到字串中間的字元。
        raw = input("請輸入剪刀(0)、石頭(1)、布(2)：").strip()
        try:
            value = int(raw)
            # 函式內可以呼叫其他函式的時機：
            # 函式定義（def）是在執行到那一行時建立函式物件並綁定名稱；
            # 只要 if __name__ == "__main__": 在最下方。
            my_int = check_my_rps(value)
            com_int = com_int = com_random()

            print(f"你出了：{CHOICES[my_int]} ({my_int})")
            print(f"電腦出了：{CHOICES[com_int]} ({com_int})")

            msg=assess_rps(my_int,com_int)
            print(msg)
            return my_int
        except ValueError as exc:

            # if str(exc)：在 Python 中，非空字串在布林判斷下視為 True，空字串 "" 視為 False。
            # str(exc)：把捕捉到的例外物件 exc 轉成字串（例如取得 raise ValueError("請輸入整數 0-2...") 裡的訊息）。
            # else "請輸入數字 0-2"：如果 str(exc) 有內容，就用 str(exc)；如果是空字串（例如有人寫 raise ValueError() 沒帶訊息），就改用預設提示 "請輸入數字 0-2"。
            msg = str(exc) if str(exc) else "請輸入數字 0-2"
            print(f"輸入無效：{msg}，請再試一次。")
```

```python
# 剪刀(0)、石頭(1)、布(2)
def assess_rps(my_int=0, com_int=0):
    if (com_int < 2 and (com_int + 1)== my_int) or (com_int == 2 and my_int == 0):
        msg = "win"
    elif (my_int < 2 and (my_int + 1)== com_int) or (my_int == 2 and com_int == 0):
        msg = "lost"
    else:
        msg = "even"
    return msg
```

```python
def com_random():
    com = random.randint(0,2)
    return com
```

```python
import random

# 使用字典形式
CHOICES = {
    0: "剪刀",
    1: "石頭",
    2: "布",
}

# 當你直接執行這個 .py 檔案時，Python 會把該檔案的 __name__ 設為 "__main__"，因此 if __name__ == "__main__": 裡面的程式會執行。
# 當另一個檔案用 import 匯入這個模組時，該模組的 __name__ 會是它的檔名（不是 "__main__"），因此這塊裡面的測試或主流程不會自動執行。
if __name__ == "__main__":
    my_int = get_player_choice()

```
