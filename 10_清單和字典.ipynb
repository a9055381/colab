```python
# 串列可以用 index 進行索引
a = [20, 30, 10, 5]

print(a[0])

# sum 總和函式、max 找最大值函式、min 找最小值函式
print(sum(a), max(a), min(a))

# len 取個數函式
print(len(a))

# 直接取串列最後一組值
print(a[-1])

# 配合 for 迴圈使用
for n in a:
    print(n)

# 索引取值方式
# 串列[開始位置：結束位置(不包含):間隔(正數為向右取，負數為向左取)]，字串也可以用
print(a[-1:-3:-1])

# 更改值
a[0]=70
print(a)

a[0:3]=[1]
print(a)

# 使用清單時機：同類型的多種東西，例如一間教室內的學生、一副撲克牌
# 建議以同類型狀態使用清單
```

```python
# 當不同類型的資料時，可以使用 dict 來表示，例如：學生的基本資料
student = {
    "name":"陳大明",
    "age":18,
    "score":90
}
print(student)

student["age"] = 20
print(student)

student["paied"] = True
print(student)

# 配合 for 迴圈使用，取出的 n 為 key 值
for n in student:
    print(n, student[n])
```

```python
a = ["a", "b", "c", "d", "e", "f", "g", "h"]
print("a：", a)
print("a[1:5]：", a[1:5])
print("a[2:-2]：", a[2:-2])
print("a[:5]：", a[:5])
print("a[2:]：", a[2:])
print("a[2:-1:-1]：", a[2:-1:-1])
print("a[-1:-3:-1]：", a[-1:-3:-1])
print("a[::-1]：", a[::-1])
```

```python
# 字串也可以用 index 進行索引
b = "apple"

print(b[0])
```
