$BMI=\frac{體重(公斤)}{身高^2(公尺)}$

```python
# 範例
bmi=78 / (175 / 100) ** 2
print(f"BMI 為 {bmi:.2f} ")
```

```python
def calc_bmi(weight, height):
    bmi = weight / (height / 100) ** 2
    return bmi
```

```python
def assess_bmi(bmi):
    if bmi > 25:
        return "體重過重"
    elif bmi > 18:
        return "健康體重"
    else:
        return "體重過輕"
```

```python
# 功能(參數)，就算沒有參數，也要保留小括號
try:
    height = int(input("請輸入您的身高(公分)："))
    weight = int(input("請輸入您的體種(公斤)："))
except ValueError:
    print("身高、體重要輸入數值哦！")
else:
    bmi = calc_bmi(weight, height)
    result = assess_bmi(bmi)
    print(f"BMI 為 {bmi:.2f}，{result}")
```
